
# Lab–15 (Python OOP) – Complete Question Set (No Duplicates)

## 1. Class Creation & Constructor
1. Write a Python program to create a class named **Students** that initializes attributes like **name, age, and grade** using a constructor, and displays the details using a method.
2. Create a class **Bank_Account** with data members `Account_No`, `User_Name`, `Email`, `Account_Type`, and `Account_Balance`. Include methods `GetAccountDetails()` and `DisplayAccountDetails()`.
3. Create a class **Circle** with methods to calculate **area and perimeter**.

## 2. Object Interaction
4. Define a class **Time** with `hour` and `minute` as data members and define a method to add two-time objects.

## 3. Inheritance
5. Demonstrate **Single Inheritance** using base class `Person` and derived class `Student`.
6. Demonstrate **Multiple Inheritance** where a class inherits properties and methods from more than one parent class.
7. Demonstrate **Multilevel Inheritance** where a class is derived from another derived class.
8. WAP to demonstrate **all types of inheritance** (Single, Multiple, Multilevel, Hierarchical, Hybrid).

## 4. Constructor & Method Concepts
9. Demonstrate **Constructor Overloading** using default arguments.
10. Demonstrate **Method Overloading** using variable-length arguments.
11. Demonstrate **Method Overriding** in inheritance.
12. Create a `Person` class with constructor (`name`, `age`) and a child class `Employee` that inherits from `Person` and adds `salary`. Override `__init__` using `super()`.

## 5. Polymorphism & Abstraction
13. Create an abstract base class **Shape** with a method `draw()`. Create child classes `Rectangle`, `Circle`, and `Triangle` that implement the `draw()` method. Create a list containing objects of each class and call `draw()` on each.
14. Illustrate **Polymorphism** through method overriding and dynamic method dispatch.

## 6. Encapsulation, Operator Overloading & Advanced OOP
15. Demonstrate **Operator Overloading** using special methods.
16. Demonstrate **Encapsulation** using private data members.
17. Demonstrate **Abstraction** using the `abc` module.

## 7. Static & Class Methods
18. Write a Python program to demonstrate the use of **Static Methods and Class Methods**.

## 8. Balance Sheet Program
19. Create a class **Profit** with a method `getProfit()` (input from user).
20. Create a class **Loss** with a method `getLoss()` (input from user).
21. Create a class **BalanceSheet** that inherits from both (Multiple Inheritance) and calculates the balance. It should have methods `getBalance()` and `printBalance()`.

---

*Note: Each program should follow structure: Aim, Algorithm, Code, Output, Conclusion.*
