# List

### 01) WAP to find sum of all the elements in a List.

```python
lst = [1,2,3,4,5]
ans = 0
for i in lst:
    ans += i
print(ans)
print(sum(lst))
```

### 02) WAP to find largest element in a List.

```python
lst = [12, 45, 67, 2]
max_no = 0
for i in lst:
    if max_no < i:
        max_no = i
print(max_no)
print(max(lst))
```

### 03) WAP to find the length of a List.

```python
lst = [3,5,7,9]
count = 0
for _ in lst:
    count += 1
print('Length =', count)
print('Length =', len(lst))
```

### 04) WAP to interchange first and last elements in a list.

```python
lst = [10,20,30,40]
lst[0], lst[-1] = lst[-1], lst[0]
print(lst)
```

### 05) WAP to split the List into two parts and append the first part to the end.

```python
lst = [1,2,3,4,5,6]
mid = len(lst)//2
new_list = lst[mid:] + lst[:mid]
print(new_list)
```

### 06) WAP to interchange the elements on two positions entered by a user.

```python
arr = [10, 20, 30, 40, 50]

print("Original array:", arr)

pos1 = int(input("Enter first position : ")) - 1
pos2 = int(input("Enter second position : ")) - 1

if 0 <= pos1 < len(arr) and 0 <= pos2 < len(arr):
    arr[pos1], arr[pos2] = arr[pos2], arr[pos1]
    print("Array after swapping:", arr)
else:
    print("Invalid positions!")
```

### 07) WAP to reverse the list entered by user.

```python
n = int(input("Enter size of list: "))

arr = []

for i in range(n):
    val = int(input(f"Enter element {i+1}: "))
    arr.append(val)

print("Original list:", arr)

reversed_list = arr[::-1]

print("Reversed list:", reversed_list)
```

### 08) WAP to print even numbers in a list.

```python
arr = list(map(int, input("Enter numbers separated by space: ").split()))

print("Even numbers in the list:")

for num in arr:
    if num % 2 == 0:
        print(num, end=" ")
```

### 09) WAP to count unique items in a list.

```python
my_list = [1, 2, 2, 3, 4, 4, 5]
unique_count = len(set(my_list))

print("Number of unique items:", unique_count)
```

### 10) WAP to copy a list.

```python
arr = [10, 20, 30, 40]
copy_arr = arr.copy()

print("Original:", arr)
print("Copy:", copy_arr)
```

### 11) WAP to print all odd numbers in a given range.

```python
arr = list(map(int, input("Enter list elements: ").split()))

start = int(input("Enter start index: "))
end = int(input("Enter end index: "))

print("Odd numbers in the given index range:")

for i in range(start, end + 1):
    if arr[i] % 2 != 0:
        print(arr[i], end=" ")
```

### 12) WAP to count occurrences of an element in a list.

```python
arr = list(map(int, input("Enter elements separated by space: ").split()))

x = int(input("Enter element to count: "))

count = 0
for num in arr:
    if num == x:
        count += 1

print(f"{x} occurs {count} time(s) in the list.")
```

### 13) WAP to find second largest number in a list.

```python
arr = list(map(int, input("Enter elements: ").split()))

if len(arr) < 2:
    print("Second largest does not exist.")
else:
    largest = float('-inf')
    second_largest = float('-inf')

    for num in arr:
        if num > largest:
            second_largest = largest
            largest = num
        elif num > second_largest and num != largest:
            second_largest = num

    if second_largest == float('-inf'):
        print("Second largest does not exist.")
    else:
        print("Second largest number is:", second_largest)
```

### 14) WAP to extract elements with frequency greater than K.

```python
arr = list(map(int, input("Enter list elements: ").split()))
k = int(input("Enter value of K: "))

freq = {}

for num in arr:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1

print(f"Elements with frequency greater than {k}:")

for key in freq:
    if freq[key] > k:
        print(key, end=" ")

# lst = [1,2,2,3,3,3,4,4]
# K = 2
# from collections import Counter
# freq = Counter(lst)
# result = [item for item, count in freq.items() if count > K]
# print(result)
```

### 15) WAP to create a list of squared numbers from 0 to 9 with and without using List Comprehension.

```python
squares = [i ** 2 for i in range(10)]
print("With List Comprehension:", squares)
```

### 16) WAP to create a new list (fruit whose name starts with 'b') from the list of fruits given by user.

```python
fruits = input("Enter fruits separated by spaces: ").split()

b_fruits = [fruit for fruit in fruits if fruit.lower().startswith('b')]

print("Fruits starting with 'b':", b_fruits)
```

### 17) WAP to create a list of common elements from given two lists.

```python
list1 = input("Enter elements of first list separated by spaces: ").split()
list2 = input("Enter elements of second list separated by spaces: ").split()

common_elements = [item for item in list1 if item in list2]

print("Common elements:", common_elements)
```
