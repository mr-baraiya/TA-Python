### 01) Write a Python program to create a class named Students that initializes attributes like name, age, and grade using a constructor, and displays the details using a method.

```
class Students:
    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade

    def display_details(self):
        print(f"Name: {self.name}, Age: {self.age}, Grade: {self.grade}")

student = Students("Vishal", 20, "A")
student.display_details()
```

### 03) Create a class Circle with methods to calculate area and perimeter.

```
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

    def perimeter(self):
        return 2 * 3.14 * self.radius

c = Circle(5)
print("Area:", c.area())
print("Perimeter:", c.perimeter())
```

### 04) Define a class Time with hour and minute as data members and define a method to add two-time objects.

```
class Time:
    def __init__(self, hour, minute):
        self.hour = hour
        self.minute = minute

    def add_time(self, t):
        total_minutes = self.minute + t.minute
        extra_hour = total_minutes // 60
        new_minute = total_minutes % 60
        new_hour = self.hour + t.hour + extra_hour
        return Time(new_hour, new_minute)

    def display(self):
        print(f"{self.hour} hour(s) and {self.minute} minute(s)")

t1 = Time(2, 50)
t2 = Time(1, 20)
t3 = t1.add_time(t2)
t3.display()
```

### 05) Demonstrate Single Inheritance using base class Person and derived class Student.

```
class Person:
    def __init__(self, name):
        self.name = name

class Student(Person):
    def __init__(self, name, grade):
        super().__init__(name)
        self.grade = grade

    def display(self):
        print(f"Name: {self.name}, Grade: {self.grade}")

s = Student("Vishal", "A")
s.display()
```
