```python
s = 'madam'
if s == s[::-1]: print('Palindrome')
else: print('Not Palindrome')

# s = input("Enter a string: ")
# cleaned = ''.join(s.split()).lower()
# if cleaned == cleaned[::-1]:
#     print("Palindrome")
# else:
#     print("Not Palindrome")
```

## 2. WAP to reverse the words in given string.

```python
s = input("Enter the String : ")
words = s.split()
rev = ' '.join(words[::-1])
print(rev)
```

## 3. WAP to remove ith character from given string.

```python
s = input("Enter the String : ")
i = int(input("Enter the number : "))
s = s[:i]+s[i+1:]
print(s)
```

## 4. WAP to find length of String without using len() function.

```python
s = input("Enter the String : ")
count = 0
for _ in s:
    count += 1
print('Length =', count)
```

## 5. WAP to print even length words in string.

```python
s = input("Enter the String : ")
for word in s.split():
    if len(word) % 2 == 0:
        print(word)
        
```

## 6. WAP to count number of vowels in given string.

```python
s = input("Enter the String : ")
vowels = 'aeiouAEIOU'
count = 0
for ch in s:
    if ch in vowels:
        count += 1
print('Vowel count =', count)
```

## 7. WAP to capitalize the first and last character of each word in a string.

```python
s = input("Enter the String : ")
res = []
for w in s.split():
    if len(w) > 1:
        w = w[0].upper() + w[1:-1] + w[-1].upper()
    else:
        w = w.upper()
    res.append(w)
print(' '.join(res))
```
