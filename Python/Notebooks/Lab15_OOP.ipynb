## 1. Write a Python program to create a class named Students that initializes attributes like name, age, and grade using a constructor, and displays the details using a method.

```
# Write your solution for Q1
```

```
class Students:
    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade
    def display(self):
        print(self.name, self.age, self.grade)

s = Students('Vishal',20,'A')
s.display()
```

## 2. Create a class Bank_Account with data members Account_No, User_Name, Email, Account_Type, and Account_Balance. Include methods GetAccountDetails() and DisplayAccountDetails().

```
# Write your solution for Q2
```

```
class Bank_Account:
    def __init__(self):
        self.Account_No='' 
        self.User_Name='' 
        self.Email='' 
        self.Account_Type='' 
        self.Account_Balance=0
    def GetAccountDetails(self):
        self.Account_No='12345'
        self.User_Name='Vishal'
        self.Email='vishal@gmail.com'
        self.Account_Type='Saving'
        self.Account_Balance=5000
    def DisplayAccountDetails(self):
        print(self.Account_No, self.User_Name, self.Email, self.Account_Type, self.Account_Balance)

b = Bank_Account()
b.GetAccountDetails()
b.DisplayAccountDetails()
```

## 3. Create a class Circle with methods to calculate area and perimeter.

```
# Write your solution for Q3
```

```
class Circle:
    def __init__(self,r): self.r=r
    def area(self): return 3.14*self.r*self.r
    def peri(self): return 2*3.14*self.r

c=Circle(5)
print('Area:',c.area(),'Perimeter:',c.peri())
```

## 4. Define a class Time with hour and minute as data members and define a method to add two-time objects.

```
# Write your solution for Q4
```

```
class Time:
    def __init__(self,h,m): self.h=h; self.m=m
    def add(self, t2):
        h = self.h + t2.h
        m = self.m + t2.m
        h += m//60
        m %= 60
        return Time(h,m)
    def display(self): print(self.h,':',self.m)

t1=Time(2,50); t2=Time(1,30)
t3=t1.add(t2)
t3.display()
```

## 13. Write a Python program to demonstrate Abstraction using the abc (Abstract Base Class) module.

```
# Write your solution for Q13
```

```
from abc import ABC, abstractmethod
class A(ABC):
    @abstractmethod
    def show(self): pass
class B(A):
    def show(self): print('Implemented')
B().show()
```

## 14. Write a Python program to illustrate Polymorphism through method overriding and dynamic method dispatch.

```
# Write your solution for Q14
```

```
class A: def show(self): print('A')
class B(A): def show(self): print('B')
def call(obj): obj.show()
call(A()); call(B())
```

## 15. Write a Python program to demonstrate the use of Static Methods and Class Methods.

```
# Write your solution for Q15
```

```
class Demo:
    @staticmethod
    def s(): print('Static')
    @classmethod
    def c(cls): print('Class Method')
Demo.s(); Demo.c()
```
