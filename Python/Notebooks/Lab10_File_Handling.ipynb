## 3. WAP to read first 5 lines from the file named abc.txt.

```
# Write your solution for Q3 here
```

```
with open('abc.txt','r') as f:
    for i in range(5):
        print(f.readline(), end='')
```

## 4. WAP to find the longest word in a file named abc.txt.

```
# Write your solution for Q4 here
```

```
with open('abc.txt','r') as f:
    words = f.read().split()
longest = max(words, key=len)
print('Longest word:', longest)
```

## 5. WAP to find the size of the file named abc.txt.

```
# Write your solution for Q5 here
```

```
import os
size = os.path.getsize('abc.txt')
print('File size:', size, 'bytes')
```

## 6. WAP to implement search function to search specific occurrence of word in a given text file.

```
# Write your solution for Q6 here
```

```
def search_word(filename, word):
    with open(filename,'r') as f:
        text = f.read()
    return text.count(word)

print(search_word('abc.txt','python'))
```
