## 4. Unique Word Analyzer — Read article.txt, extract words, make set for unique words, create frequency dictionary, store result in wordcount.txt.

```
# Write your solution for Q4 here
```

```
try:
    with open('article.txt', 'r') as f:
        text = f.read()
    words = text.split()
    unique_words = set(words)
    freq = {}
    for w in words:
        freq[w] = freq.get(w, 0) + 1
    with open('wordcount.txt', 'w') as f:
        for k,v in freq.items():
            f.write(f"{k}: {v}\n")
    print('wordcount.txt created successfully')
except FileNotFoundError:
    print('article.txt not found')
```
