## 1. WAP to handle Divide By Zero Exception.

```
# Write your solution for Q1 here
```

```
try:
    a = 10
    b = 0
    print(a/b)
except ZeroDivisionError:
    print('Error: Cannot divide by zero')
```

## 2. WAP to handle File Not Found Exception.

```
# Write your solution for Q2 here
```

```
try:
    f = open('unknown.txt','r')
except FileNotFoundError:
    print('Error: File not found')
```

## 3. WAP to handle Type Exception.

```
# Write your solution for Q3 here
```

```
try:
    x = 'hello' + 5
except TypeError:
    print('Error: Type mismatch')
```

## 4. WAP to demonstrate ValueError and IndexError with example.

```
# Write your solution for Q4 here
```

```
try:
    num = int('abc')
except ValueError:
    print('ValueError caught')

try:
    lst = [1,2,3]
    print(lst[5])
except IndexError:
    print('IndexError caught')
```
