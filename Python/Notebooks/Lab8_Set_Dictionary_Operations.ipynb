```
# Write your solution for Q3 here
```

## 5. WAP to check if two lists have at least one element common.

```
# Write your solution for Q5 here
```

```
l1 = [1,2,3]
l2 = [3,4,5]
print(len(set(l1) & set(l2)) > 0)
```

## 6. WAP to sort dictionary by key or value.

```
# Write your solution for Q6 here
```

```
d = {'b':2,'a':1,'c':3}
print('Sort by key:', dict(sorted(d.items())))
print('Sort by value:', dict(sorted(d.items(), key=lambda x: x[1])))
```

## 7. WAP to find the sum of all items in a dictionary given by user.

```
# Write your solution for Q7 here
```

```
d = {'a':10,'b':20,'c':30}
print('Sum:', sum(d.values()))
```

## 8. WAP to check if a given string is binary string or not.

## 9. WAP to find common elements in three lists using set.

```
# Write your solution for Q9 here
```

```
l1 = [1,2,3]
l2 = [2,3,4]
l3 = [3,2,5]
print(set(l1) & set(l2) & set(l3))
```

## 10. WAP to merge two dictionaries given by user.

```
# Write your solution for Q10 here
```

```
d1 = {'a':1,'b':2}
d2 = {'c':3}
res = d1.copy()
res.update(d2)
print(res)
```

## 11. WAP to count number of vowels in given string using set.

```
# Write your solution for Q11 here
```

```
s = 'Beautiful Day'
vowels = set('aeiouAEIOU')
print(sum(1 for ch in s if ch in vowels))
```
