## 1. WAP to count simple interest using function.

```
# Write your solution for Q1 here
```

```
def simple_interest(P, R, T):
    return (P * R * T) / 100
print(simple_interest(1000,5,2))
```

## 3. WAP to find maximum number from given two numbers using function.

```
# Write your solution for Q3 here
```

```
def max_of_two(a,b):
    return a if a>b else b
print(max_of_two(10,20))
```

## 4. WAP that defines a function which returns 1 if the number is prime otherwise return 0.

```
# Write your solution for Q4 here
```

```
def is_prime(n):
    if n<2: return 0
    for i in range(2,int(n**0.5)+1):
        if n%i==0:
            return 0
    return 1
print(is_prime(17))
```

## 5. WAP to generate Fibonacci series of N given number using function name fibbo.

```
# Write your solution for Q5 here
```

```
def fibbo(n):
    a,b=0,1
    for _ in range(n):
        print(a, end=' ')
        a,b=b,a+b
fibbo(7)
```

## 6. WAP to find the factorial of a given number using recursion.

```
# Write your solution for Q6 here
```

```
def fact(n):
    if n==0: return 1
    return n*fact(n-1)
print(fact(5))
```

## 7. WAP to implement simple calculator using lambda function.

```
# Write your solution for Q7 here
```

```
add = lambda a,b: a+b
sub = lambda a,b: a-b
mul = lambda a,b: a*b
div = lambda a,b: a/b
print(add(5,3), sub(5,3), mul(5,3), div(6,3))
```
