
# Lab 1 — Python Setup and First Program

## 1. Install latest version of Python
### Steps:
- Visit the official Python website.
- Download the latest **Python 3.x** version.
- Run the installer.
- Make sure to check **“Add Python to PATH”**.
- Verify installation:
```bash
python --version
```

## 2. Install Anaconda to use Jupyter Notebook
### Steps:
- Go to the Anaconda distribution page.
- Download **Anaconda Individual Edition**.
- Install with default settings.
- Launch **Jupyter Notebook** from Anaconda Navigator.

## 3. Run “Hello World” using different methods

### a) Python IDLE — Interactive Mode
```python
>>> print("Hello World")
```

### b) Python IDLE — Script Mode
Save a file `hello.py` with:
